```python
%sql

create catalog if not exists deltatable;
use catalog deltatable;
create schema if not exists sample_df;
use schema sample_df;
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("LargeSampleDF").getOrCreate()

data = [
    ("Alice", 15, 10, "A"),
    ("Bob", 16, 11, "B"),
    ("Charlie", 14, 9, "A"),
    ("David", 17, 12, "C"),
    ("Eva", 15, 10, "B"),
    ("Frank", 16, 11, "A"),
    ("Grace", 14, 9, "A"),
    ("Hannah", 17, 12, "B"),
    ("Ian", 15, 10, "C"),
    ("Julia", 16, 11, "A")
]

columns = ["Name", "Age", "Grade", "Section"]

df = spark.createDataFrame(data, columns)
df.show(truncate=False)
```

```python
df.write.mode("overwrite").format("delta").save("/Volumes/deltatable/sample_df/delta/deltasink/")
```
