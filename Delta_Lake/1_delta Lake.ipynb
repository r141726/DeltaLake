```python
df = spark.read.table('deltatable.sales_orders.sales_data')


display(df)
```

```python
from pyspark.sql.functions import col

df = df.withColumnRenamed("Order Date","Order_Date")\
    .withColumnRenamed("Order ID","Order_ID")\
    .withColumnRenamed("Purchase Address","Purchase_Address")\
    .withColumnRenamed("Quantity Ordered","Quantity_Ordered")\
    .withColumnRenamed("Price Each","Price_Each")\
    .withColumnRenamed("Cost price","Cost_price")

display(df)
```

```python
df.write.format("delta").mode("overwrite").option('mergeSchema',True).save("/Volumes/deltatable/sales_orders/sales_orders_dump")
```
